This notebook builds a complete multi-qubit chip from start to finish: lay it out, analyze and tune it, and export it to GDS for fabrication.

The chip mixes the parts a real design uses: transmon pockets and crossmons with flux lines, a tunable coupler, meandered and straight transmission lines, capacitive couplers, a multiplexed readout line, and launchpads for the wire bonds. It is a teaching layout rather than a working device -- the resonator lengths are only roughly chosen -- but every step is the one you would take on a real chip.

**What you'll do**

1. **Layout** -- place the qubits, connect them with buses, add readout and control lines.
2. **Analyze** -- extract the capacitance matrix and run the lumped-oscillator model (LOM) with Ansys Q3D, then an eigenmode simulation and energy-participation (EPR) analysis with HFSS.
3. **Export** -- write the mask to GDS.

#### The Busses

We now couple the qubits to each other, primarily using RouteMeander. Although one needs to run simulations to properly tune the line lengths for target frequencies, an initial estimate could be determined from the below method;

```python
from qiskit_metal.analyses.em.cpw_calculations import guided_wavelength


def find_resonator_length(frequency, line_width, line_gap, N):
    # frequency in GHz
    # line_width/line_gap in um
    # N -> 2 for lambda/2, 4 for lambda/4

    [lambdaG, etfSqrt, q] = guided_wavelength(
        frequency * 10**9,
        line_width * 10**-6,
        line_gap * 10**-6,
        750 * 10**-6,
        200 * 10**-9,
    )
    return str(lambdaG / N * 10**3) + " mm"
```

#### Without Ansys: capacitance with ElmerFEM

The Q3D run above needs Ansys. The open-source path computes the same capacitance matrix with gmsh (mesh) and [ElmerFEM](https://www.elmerfem.org/) (electrostatics), and LOM 2.0 turns it into qubit parameters; tutorial [4.19](../../tut/4-Analysis/4.19-Analyze-a-transmon-using-ElmerFEM.ipynb) walks through each step on a single transmon.

The gmsh and Elmer renderers read a layer stack, which a `MultiPlanar` design carries, so `Q_Main` is copied, with the same options, into a small design of its own: a 2 mm × 2 mm chip centered on the qubit. As in the Q3D run, every connection pad is left open. This needs `pip install "quantum-metal[mesh]"` and `ElmerSolver` on your `PATH`; without them the cells below say so and do nothing.

```python
import copy
import shutil
import tempfile

HAVE_ELMER = shutil.which("ElmerSolver") is not None
if not HAVE_ELMER:
    print("ElmerSolver not found: install ElmerFEM (https://www.elmerfem.org/) to run this section.")
else:
    import gmsh
    from qiskit_metal.renderers.renderer_elmer.elmer_renderer import QElmerRenderer

    q_main_cell = designs.MultiPlanar({}, True)
    q_main_cell.chips.main.size.update(
        size_x="2mm", size_y="2mm", center_x=q_main.options.pos_x, center_y=q_main.options.pos_y
    )
    TransmonPocket6(q_main_cell, "Q_Main", options=copy.deepcopy(q_main.options))

    elmer = QElmerRenderer(q_main_cell)
    if not gmsh.isInitialized():
        gmsh.initialize()
    gmsh.option.setNumber("General.Terminal", 0)
    elmer.options.simulation_dir = tempfile.mkdtemp(prefix="q_main_")  # Elmer's scratch files
    elmer.gmsh.options.mesh.min_size = "5um"
    elmer.gmsh.options.mesh.max_size = "50um"
    elmer.render_design(open_pins=[("Q_Main", pin) for pin in q_main.pins], skip_junctions=True)
    elmer.add_solution_setup("capacitance")
    elmer.run("capacitance")
    elmer_cmat = elmer.capacitance_matrix  # fF
    elmer.close()
    display(elmer_cmat.round(2))
```

The rows match the Q3D matrix above, named differently (`Q_Main_pad_top` for `pad_top_Q_Main`); the `ground_plane` diagonal is a placeholder, because ground is the reference conductor. The values are a little larger than Q3D's, and they come down as the mesh is refined (they also change by about 0.5% from one run to the next, because the mesh does):

| mesh `min_size` / `max_size` | readout pad | `pad_top` | `pad_top`–`pad_bot` | solve time |
|---|---|---|---|---|
| 5 um / 50 um (this cell) | 53.6 fF | 105.8 fF | −34.3 fF | ~1 min |
| 3 um / 30 um | 48.4 fF | 100.8 fF | −32.7 fF | ~8 min |
| Q3D above (16 passes) | 42.1 fF | 97.5 fF | −31.2 fF | — |

For tuning, check convergence the same way as with Q3D: refine the mesh until the entries you care about stop moving.

#### LOM 2.0 on the Elmer matrix

LOM 2.0 builds the circuit from the matrix: the junction (14 nH, 2 fF, as in `c1.setup`) between the two pads makes the transmon, and each connection pad ends a transmission-line resonator at the frequency used above.

```python
if HAVE_ELMER:
    from scipy.constants import speed_of_light
    from qiskit_metal.analyses.quantization.lom_core_analysis import (
        Cell,
        CompositeSystem,
        Subsystem,
    )

    line_freqs = {"readout": 7.0, "bus_01": 5.6, "bus_02": 5.7, "bus_03": 5.8, "bus_04": 5.9, "bus_05": 6.0}
    junction = ("Q_Main_pad_top", "Q_Main_pad_bot")
    cell = Cell(
        dict(
            node_rename={f"Q_Main_{pad}_connector_pad": pad for pad in line_freqs},
            cap_mat=elmer_cmat,
            ind_dict={junction: 14},  # nH
            jj_dict={junction: "j1"},
            cj_dict={junction: 2},  # fF
        )
    )
    qubit = Subsystem(name="Q_Main", sys_type="TRANSMON", nodes=["j1"])
    lines = [
        Subsystem(
            name=pad,
            sys_type="TL_RESONATOR",
            nodes=[pad],
            q_opts=dict(f_res=f, Z0=50, vp=0.404314 * speed_of_light),
        )
        for pad, f in line_freqs.items()
    ]
    system = CompositeSystem(
        subsystems=[qubit, *lines],
        cells=[cell],
        grd_node="ground_plane",
        nodes_force_keep=list(line_freqs),
    )
    results = system.hamiltonian_results(system.add_interaction(), print_info=False)
    chi = results["chi_in_MHz"].to_dataframe()
    g = system.compute_gs().to_dataframe()

    print(f"f_Q   = {results['fQ_in_Ghz']['Q_Main']:.3f} GHz")
    print(f"alpha = {chi.loc['Q_Main', 'Q_Main']:.0f} MHz")
    print("\n           g (MHz)   chi (MHz)")
    for pad in line_freqs:
        print(f"{pad:9s} {abs(g.loc['j1', pad]):8.1f}   {chi.loc['Q_Main', pad]:8.2f}")
```

The Q3D matrix with LOM 1 above gave 4.96 GHz and −351 MHz, with couplings of 44–52 MHz. The Elmer numbers are lower in frequency and anharmonicity because the coarser Elmer mesh gives the pads more capacitance: a larger total capacitance lowers the charging energy $E_C$, and with it both the frequency and the anharmonicity.
